Proxies are:

engineered features that approximate real-world factors you don’t directly observe in the dataset

We don’t have:
warehouse congestion
staffing levels
shipping delays
backlog queues

the following  simulate them using data you do have (sales, temperature, etc.)

Sales Velocity represents Rate of change in demand:
    Sudden spikes → require rapid staffing (VET)
    Sudden drops → excess labor (VTO)
Example:
    Sales jump from 10K → 15K
    → velocity = +50%
    → signal: demand surge

Backlog Proxy represents Demand pressure relative to recent trend- Are we falling behind or catching up?
    If demand > recent average → system under stress
    If demand < average → slack capacity

Extreme Temperature represents Environmental stress on operations - External conditions making operations harder
    Extreme heat/cold:
        slows workers
        affects logistics
        increases delays

```python
# the following proxies give  positive → demand spike negative → demand drop

# Rate of change in demand
df['sales_velocity'] = df['Weekly_Sales'].pct_change()

df['backlog_proxy'] = (
    df['Weekly_Sales'] - df['Weekly_Sales'].rolling(4).mean()
) / df['Weekly_Sales'].rolling(4).mean()

high_temp = df['Temperature'].quantile(0.90)
low_temp  = df['Temperature'].quantile(0.10)

df['extreme_temp'] = (
    (df['Temperature'] > high_temp) |
    (df['Temperature'] < low_temp)
).astype(int)

# clean NaNs
df = df.dropna()

exog = pd.concat([exog, df[['sales_velocity', 'backlog_proxy', 'extreme_temp']]], axis=1)
```

## Predict future data

For the next  12 weeks:
    
    Predicted demand
    VET / VTO decision
    Estimated staffing impact
    Estimated cost

This turns the project into a decision engine

```python
future_pred = forecaster.predict(
    steps= 12,
    exog  = X_test
)

#preds.head()
```

```python
future_df = pd.DataFrame({
    'predicted_demand': future_pred
})
```

```python
# LOW -------- NORMAL -------- HIGH
#     25%        50%        75%

lower_threshold = y_train.quantile(0.25)
upper_threshold = y_train.quantile(0.75)

print("Lower threshold (VTO):", lower_threshold)
print("Upper threshold (VET):", upper_threshold)
```

```python
# Apply VET/VTO logic
def demand_to_action(value, lower_thr, upper_thr):
    if value < lower_thr:
        return 'VTO'
    elif value > upper_thr:
        return 'VET'
    else:
        return 'NORMAL'

future_df['decision'] = future_df['predicted_demand'].apply(
    lambda x: demand_to_action(x, lower_threshold, upper_threshold)
)
```

### Estimate staffing impact 

```python
PRODUCTIVITY_PER_WORKER = 5000  # adjust if needed - How many units of Weekly_Sales one worker can handle per week

future_df['extra_workers_needed'] = (
    (future_df['predicted_demand'] - upper_threshold)
    / PRODUCTIVITY_PER_WORKER
).clip(lower=0)

future_df['workers_to_reduce'] = (
    (lower_threshold - future_df['predicted_demand'])
    / PRODUCTIVITY_PER_WORKER
).clip(lower=0)

future_df[['extra_workers_needed', 'workers_to_reduce']] = \
    future_df[['extra_workers_needed', 'workers_to_reduce']].round(0)
```

Estimate future cost impact

```python
future_df['estimated_cost'] = 0.0
# Define wages assumptions
hourly_wage = 20
overtime_wage = 30

# VET cost
mask_vet = future_df['decision'] == 'VET'

future_df.loc[mask_vet, 'estimated_cost'] = (
    future_df.loc[mask_vet, 'extra_workers_needed']
    * overtime_wage
)

# VTO savings (negative cost)
mask_vto = future_df['decision'] == 'VTO'

future_df.loc[mask_vto, 'estimated_cost'] = (
    - future_df.loc[mask_vto, 'workers_to_reduce']
    * hourly_wage
)
```

```python
future_df['cumulative_future_cost'] = future_df['estimated_cost'].cumsum()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(future_df.index, future_df['predicted_demand'], marker='o')

plt.axhline(upper_threshold, linestyle='--', label='VET Threshold')
plt.axhline(lower_threshold, linestyle='--', label='VTO Threshold', color='orange')

plt.title('Future Demand Forecast with VET/VTO Decisions')
plt.xlabel('Date')
plt.ylabel('Predicted Demand')

plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.plot(future_df.index, future_df['cumulative_future_cost'], marker='o')

plt.axhline(0, linestyle='--')

plt.title('Projected Staffing Cost Impact (Next 12 Weeks)')
plt.ylabel('Cumulative Cost ($)')

plt.show()
```
