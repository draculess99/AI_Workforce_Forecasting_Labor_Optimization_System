# Load data

We will use the electricity demand dataset found [here](https://github.com/tidyverts/tsibbledata/tree/master/data-raw/vic_elec/VIC2015).

**Citation:**

Godahewa, Rakshitha, Bergmeir, Christoph, Webb, Geoff, Hyndman, Rob, & Montero-Manso, Pablo. (2021). Australian Electricity Demand Dataset (Version 1) [Data set]. Zenodo. https://doi.org/10.5281/zenodo.4659727

**Description of data:**

A description of the data can be found [here](https://rdrr.io/cran/tsibbledata/man/vic_elec.html). The data contains electricity demand in Victoria, Australia, at 30 minute intervals over a period of 12 years, from 2002 to early 2015. There is also the temperature in Melbourne at 30 minute intervals and public holiday dates.

```python
import pandas as pd

# 1. Load the data
train = pd.read_csv('train.csv')
features = pd.read_csv('features.csv')
stores = pd.read_csv('stores.csv')

# 2. Merge dataframes combines  sales (target), external factors (future use), store info
# Why: So the model has all context (Store size, Temperature, etc.) in one row.
df = train.merge(features, on=['Store', 'Date', 'IsHoliday'], how='left')
df = df.merge(stores, on=['Store'], how='left')



# 3. Handle Dates- enables sorting, time operations and rolling window
# Why: Models can't "read" '2010-02-05'. They need numbers like Month: 2, Week: 5.
df['Date'] = pd.to_datetime(df['Date'])
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.isocalendar().week
df["day"] = df['Date'].dt.dayofweek

# 4. Fill Missing Values
# Why: MarkDowns are NaN when there is no discount. Filling with 0 tells the model "No discount."
markdown_cols = ['MarkDown1', 'MarkDown2', 'MarkDown3', 'MarkDown4', 'MarkDown5']
df[markdown_cols] = df[markdown_cols].ffill().fillna(0)

df['CPI'] = df['CPI'].fillna(df['CPI'].mean())
df['Unemployment'] = df['Unemployment'].fillna(df['Unemployment'].mean())

# 5. Categorical Encoding
# Why: Computers only understand numbers, not "Type A" or "Type B".
df['Type'] = df['Type'].map({'A': 1, 'B': 2, 'C': 3})
df['IsHoliday'] = df['IsHoliday'].astype(int)

# 6. Create series_id for global model
df['series_id'] = df['Store'].astype(str) + '_' + df['Dept'].astype(str)

# 6. Sort the data (VERY IMPORTANT)
df = df.sort_values(['Store','Dept','Date'])

# 8. Keep omly needed columns
df = df[['Store','Dept','Date','Weekly_Sales','IsHoliday','series_id',
         'Temperature','Fuel_Price','CPI','Unemployment','Type','Size','Year','Month', 'day','Week']]

print(df.head())
```

## Lag features

We shift pass values of the time series forward. We discussed these 3 features in a previous notebook / video.

```python
df[f"lag_1"] = df["Weekly_Sales"].shift(periods=1)

freq = "4"
df[f"lag_{freq}"] = df["Weekly_Sales"].shift(periods=4)

freq = "3"
df[f"lag_{freq}"] = df["Weekly_Sales"].shift(periods=3)

freq = "2"
df[f"lag_{freq}"] = df["Weekly_Sales"].shift(periods=2)

freq = "52"
df[f"lag_{freq}"] = df["Weekly_Sales"].shift(periods=52)
df.head()
```

## Window features

We aggregate values within windows in the past.  We discussed these windows in a previous notebook / video.

```python
# We'll use a 4 and 13 window size.

result = (
    df["Weekly_Sales"]
    .rolling(window=4) # Pick window size.
    .agg(["mean", "std"]) # Pick statistics.
    .shift(periods=1) # Lag by 1 week to avoid data leakage.
)  

result = result.add_prefix("rolling_4_")

# add features to main dataframe

df = df.merge(result, how="left", left_index=True, right_index=True)


result = (
    df["Weekly_Sales"]
    .rolling(window=13) # Pick window size.
    .agg(["mean", "std"]) # Pick statistics.
    .shift(periods=1) # Lag by 1 week to avoid data leakage.
)  

result = result.add_prefix("rolling_13_")

# add features to main dataframe

df = df.merge(result, how="left", left_index=True, right_index=True)

import pandas as pd
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

df.head(10)
```

```python
df = df.groupby(df.index)['Weekly_Sales'].sum().to_frame()
```

## Lasso

We'll split the data into train and test.

We train the model on the train set and evaluate it on the test set.

```python
print(df.columns)
```

```python
# Split into train and test

# We leave 2012 in the test set

features = [
    'lag_1','lag_2','lag_3','lag_4','lag_52',
    'rolling_4_mean','rolling_13_mean',
    'rolling_4_std','rolling_13_std',
    'Month','Week'
]


X = df[features]
y = df['Weekly_Sales']

df = df.dropna()

split_date = '2012-01-01'

train = df[df.index < split_date]
test = df[df.index >= split_date]

X_train = train[features]
y_train = train['Weekly_Sales']

X_test = test[features]
y_test = test['Weekly_Sales']
```

```python
# lasso = Lasso(random_state=9)

# lasso.fit(X_train, y_train)
```

```python


# fit
lasso = Lasso(alpha=0.1, max_iter=5000, random_state=9)
lasso.fit(X_train, y_train)

# predict on scaled test data
preds = lasso.predict(X_test)

# convert to Series with correct index
preds = pd.Series(preds, index=y_test.index)

```

```python
#preds = lasso.predict(X_test)
preds = lasso.predict(X_test)

rmse = root_mean_squared_error(y_test, preds)

print(f"performance of lasso = {np.round(rmse, 0)}")
```

These results is better than any of the basic forecasts that we did in the previous notebook!

Same date → multiple values

There are  multiple rows per date → you must aggregate AND switch your pipeline to use it

```python
# Aggregate
df_agg = df.groupby(df.index)['Weekly_Sales'].sum().to_frame()

# Add time features
df_agg['month'] = df_agg.index.month
df_agg['week'] = df_agg.index.isocalendar().week.astype(int)

# Add lags
df_agg['lag_1'] = df_agg['Weekly_Sales'].shift(1)
df_agg['lag_2'] = df_agg['Weekly_Sales'].shift(2)
df_agg['lag_3'] = df_agg['Weekly_Sales'].shift(3)
df_agg['lag_4'] = df_agg['Weekly_Sales'].shift(4)

# Drop NaN
df_agg = df_agg.dropna()
```

```python
split_date = '2012-01-01'

train = df_agg[df_agg.index < split_date]
test = df_agg[df_agg.index >= split_date]
```

```python
features = ['lag_1', 'lag_4', 'month', 'week']

X_train = train[features]
y_train = train['Weekly_Sales']

X_test = test[features]
y_test = test['Weekly_Sales']

lasso.fit(X_train, y_train)

preds = lasso.predict(X_test)
preds = pd.Series(preds, index=y_test.index)
```

```python
# plot predictions vs actuals
fig, ax = plt.subplots(figsize=(12,5))

y_train[-100:].plot(ax=ax, label='Train')
y_test.plot(ax=ax, label='Test')
preds.plot(ax=ax, label='Prediction')

plt.title("LASSO Forecast (Aggregated)")
plt.xlabel("Date")
plt.ylabel("Weekly Sales")

plt.legend()
plt.grid()
```

The data was aggregated to a single time series to ensure one observation per date, enabling proper forecasting and visualization.”

```python
# We can also understand what is driving the predictions of this
# lasso regression.

importance = lasso.coef_
features = lasso.feature_names_in_

pd.Series(importance, index=features).sort_values(ascending=False).plot.bar()
plt.title("Feature importance")
plt.ylabel("Coefficient value")
```

The hour and the day are the most important predictors of energy consumption.

```python
y.groupby(X["day"]).mean().plot(figsize=(8,4))

plt.title("Daily energy demand")
plt.ylabel("Energy demand")
plt.xlabel("day of the week")
plt.tight_layout()
```

## Random forest

```python
rf = RandomForestRegressor(n_estimators=10, random_state=9)

rf.fit(X_train, y_train)
```

```python
preds = rf.predict(X_test)

rmse = root_mean_squared_error(y_test, preds)

print(f"performance of random forests = {np.round(rmse, 0)}")
```

These results are even better than those from Lasso!

```python
# plot predictions vs actuals

fig, ax = plt.subplots(figsize=(6, 3))
y_train[-100:].plot(ax=ax, label='train')
y_test[:100].plot(ax=ax, label='test')
preds.iloc[:100].plot(ax=ax, label='predictions')
plt.title("Random forests forecasting")
ax.legend(bbox_to_anchor=(1.3, 1.0));
```

```python
importance = rf.feature_importances_
features = rf.feature_names_in_

pd.Series(importance, index=features).sort_values(ascending=False).plot.bar()
plt.title("Feature importance")
plt.ylabel("Feature importance")
```

For the random forests, the energy consumption in the previews hour, seems to be the most important feature to predict demand.
