# Load data

We will use the electricity demand dataset found [here](https://github.com/tidyverts/tsibbledata/tree/master/data-raw/vic_elec/VIC2015).

**Citation:**

Godahewa, Rakshitha, Bergmeir, Christoph, Webb, Geoff, Hyndman, Rob, & Montero-Manso, Pablo. (2021). Australian Electricity Demand Dataset (Version 1) [Data set]. Zenodo. https://doi.org/10.5281/zenodo.4659727

**Description of data:**

A description of the data can be found [here](https://rdrr.io/cran/tsibbledata/man/vic_elec.html). The data contains electricity demand in Victoria, Australia, at 30 minute intervals over a period of 12 years, from 2002 to early 2015. There is also the temperature in Melbourne at 30 minute intervals and public holiday dates.

```python
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error
from sklearn.metrics import root_mean_squared_error

from skforecast.recursive import ForecasterRecursive
from skforecast.preprocessing import RollingFeatures


# 1. Load the data
train = pd.read_csv('train.csv')
features = pd.read_csv('features.csv')
stores = pd.read_csv('stores.csv')

# 2. Merge dataframes combines  sales (target), external factors (future use), store info
# Why: So the model has all context (Store size, Temperature, etc.) in one row.
df = train.merge(features, on=['Store', 'Date', 'IsHoliday'], how='left')
df = df.merge(stores, on=['Store'], how='left')
```

```python
#df['Date'] = pd.to_datetime(df['Date'])

# df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
#df = df.dropna(subset=['Date'])

#df_agg = df.reset_index(drop=True)   # remove fake 1970 index
df['Date'] = pd.to_datetime(df['Date'])  # use real dates

# 3. Set index
df = df.set_index('Date')

# 4. aggregated ALL stores into one time series earlier  and then created lag features on that aggregated data
df_agg = df.groupby(df.index)['Weekly_Sales'].sum().to_frame()

df_agg = df_agg.reset_index()
df_agg = df_agg.sort_index()

df = df_agg[['Date', 'Weekly_Sales']]

print(df.head())
```

## Recursive forecaster

We'll split the data into train and test.

We train the model on the train set and evaluate it on the test set.

```python
from xgboost import XGBRegressor

model = XGBRegressor(
        n_estimators=500,
        max_depth=4,
        learning_rate=0.03,
        subsample=0.8,
        colsample_bytree=0.8,
        min_child_weight=3,
        gamma=0.1,
        reg_alpha=0.1,
        reg_lambda=1.0,
        random_state=9
)
```

```python
rolling_features = RollingFeatures(
    stats        = ['mean', 'mean', 'std', 'std'],
    window_sizes = [3, 24, 3, 24]
)
```

```python
forecaster = ForecasterRecursive(
    estimator       = model,
    lags = [1, 2, 3, 4, 13],
    window_features = rolling_features
)
```

```python
# Plot the forecast vs the actual

fig, ax = plt.subplots(figsize=(8, 4))
import matplotlib.ticker as mtick
plt.gca().yaxis.set_major_formatter(mtick.StrMethodFormatter('{x:,.0f}'))

seasonal_naive_full = df['Weekly_Sales'].shift(52)
seasonal_naive_pred = seasonal_naive_full.loc[y_test.index]

seasonal_naive_pred.plot(ax=ax, label='Seasonal Naive', linestyle='--')

y_train['Weekly_Sales'].tail(100).plot(ax=ax, label='Train')
y_test['Weekly_Sales'].plot(ax=ax, label='Test')

preds.plot(ax=ax, label='XGBoost prediction')

ax.set_title("XGBoost recursive forecasting")
ax.set_xlabel("Date")
ax.set_ylabel("Weekly Sales")

ax.legend()
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import pandas as pd

# --- Seasonal naive baseline ---
seasonal_naive_full = df['Weekly_Sales'].shift(52)
seasonal_naive_pred = seasonal_naive_full.loc[y_test.index]

# --- Convert EVERYTHING to DataFrame (key fix) ---
y_train_plot = y_train.tail(100).rename(columns={'Weekly_Sales': 'Train'})
y_test_plot = y_test.rename(columns={'Weekly_Sales': 'Test'})

preds_plot = preds.to_frame(name='Recurive XGBoost Forecast')  # FIX
seasonal_plot = seasonal_naive_pred.to_frame(name='Seasonal Naive')  # FIX

# --- Create plot ---
fig, ax = plt.subplots(figsize=(10, 5))

# --- Plot ---
y_train_plot.plot(ax=ax, color='blue', linewidth=2)
y_test_plot.plot(ax=ax, color='orange', linewidth=2)
preds_plot.plot(ax=ax, color='red', linewidth=2)
seasonal_plot.plot(ax=ax, color='green', linestyle='--', linewidth=2)

# --- Zoom ---
ax.set_xlim(pd.Timestamp('2011-07-01'), None)

# --- Test split line ---
ax.axvline(x=y_test.index[0], color='black', linestyle='--', linewidth=1)

# --- Format ---
ax.yaxis.set_major_formatter(mtick.StrMethodFormatter('{x:,.0f}'))

# --- Labels ---
ax.set_title("Recursive XGBoost vs Seasonal Naive Forecast (from July 2011)", fontsize=14)
ax.set_xlabel("Date")
ax.set_ylabel("Weekly Sales")

# --- Legend ---
ax.legend()

plt.tight_layout()
plt.show()
```

Observation: the XGBoost model is overfitting seasonal lag structure (lag_52) but not learning real demand variability

-Early predictions → closer to test
-Later predictions → diverge into pattern
-Seasonal naive sometimes performs better


The XGBoost model captures seasonal patterns but produces smooth oscillations, indicating over-reliance on lag-based features. The seasonal naive baseline performs competitively and often tracks actual demand more closely, suggesting limited improvement from the machine learning mode

The recursive forecasting approach causes the model to rely increasingly on its own predictions as inputs. As a result, errors accumulate over time and the model loses responsiveness to actual demand fluctuations. This leads to smooth, repeating oscillations in the forecast, reflecting learned seasonal patterns rather than real-world variability.

```python
rmse_xgb = np.sqrt(mean_squared_error(y_test, preds))
print(rmse_xgb)
```

```python
rmse_naive = np.sqrt(mean_squared_error(y_test, seasonal_naive_pred))

print(rmse_naive)
```

```python
rmse = np.sqrt(mean_squared_error(y_test.loc[preds.index], preds))
print(rmse)
```

Conclusion: The seasonal naive baseline significantly outperforms the XGBoost model, achieving a much lower RMSE (~1.99M vs ~7.54M). This indicates that the machine learning model fails to improve upon a simple seasonal benchmark and struggles to capture actual demand variability.

The poorer performance of the XGBoost model is primarily due to the recursive forecasting approach, where predictions are fed back as inputs. This leads to error accumulation and causes the model to converge toward smooth seasonal patterns rather than adapting to real fluctuations in the data.
