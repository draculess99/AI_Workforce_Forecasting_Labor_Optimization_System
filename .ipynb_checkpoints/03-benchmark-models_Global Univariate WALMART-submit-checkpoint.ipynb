# Load data

We will use the electricity demand dataset found [here](https://github.com/tidyverts/tsibbledata/tree/master/data-raw/vic_elec/VIC2015).

**Citation:**

Godahewa, Rakshitha, Bergmeir, Christoph, Webb, Geoff, Hyndman, Rob, & Montero-Manso, Pablo. (2021). Australian Electricity Demand Dataset (Version 1) [Data set]. Zenodo. https://doi.org/10.5281/zenodo.4659727

**Description of data:**

A description of the data can be found [here](https://rdrr.io/cran/tsibbledata/man/vic_elec.html). The data contains electricity demand in Victoria, Australia, at 30 minute intervals over a period of 12 years, from 2002 to early 2015. There is also the temperature in Melbourne at 30 minute intervals and public holiday dates.

```python


# 1. Load the data
train = pd.read_csv('train.csv')
features = pd.read_csv('features.csv')
stores = pd.read_csv('stores.csv')

# 2. Merge dataframes combines  sales (target), external factors (future use), store info
# Why: So the model has all context (Store size, Temperature, etc.) in one row.
df = train.merge(features, on=['Store', 'Date', 'IsHoliday'], how='left')
df = df.merge(stores, on=['Store'], how='left')
```

```python
#df['Date'] = pd.to_datetime(df['Date'])

# df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
#df = df.dropna(subset=['Date'])

#df_agg = df.reset_index(drop=True)   # remove fake 1970 index
df['Date'] = pd.to_datetime(df['Date'])  # use real dates

# 3. Set index
df = df.set_index('Date')

# 4. aggregated ALL stores into one time series earlier  and then created lag features on that aggregated data
df_agg = df.groupby(df.index)['Weekly_Sales'].sum().to_frame()


df_agg = df_agg.reset_index()
df = df_agg[['Date', 'Weekly_Sales']]

print(df.head())
```

## Naive forecasting

We predict the next value as its immediate value in the past.

```python
df['naive_pred'] = df['Weekly_Sales'].shift(1)

df.head()
```

```python
df['lag2'] = df['Weekly_Sales'].shift(2)
df['lag3'] = df['Weekly_Sales'].shift(3)
df['lag4'] = df['Weekly_Sales'].shift(4)
df['lag13'] = df['Weekly_Sales'].shift(13)
```

We are doing single step ahead forecasting. In this scenario, we know the value in the past at the point of forecast.

## Seasonal naive forecasting

We predict the next value as its value a season ago. We know that the time series shows weekly seasonality. That means that the energy consumption each Tuesday at 9 am is more or less the same.

We can predict next Tuesday at 9 am using last Tuesday's value at 9 am

```python
# Predict sales as its value exactly 32 weeks behind.

#baseline table

# | Model          | Description         |
# | -------------- | ------------------- |
# | Naive          | last week           |
# | Moving Avg     | last 4 weeks avg    |
# | Seasonal Naive | same week last year |

# Seasonality = 52 weeks
df['seasonal_naive_pred'] = df['Weekly_Sales'].shift(52)

df.head()
```

## Moving average

We predict the next value as the average of the last 3 values.

I don't think this is a good method for this particular time series, but we'll do it just to demo how it works.

```python
# We'll use a 4H window size.

# uses last 4 weeks, #computes average, Lag by 1 wek to avoid data leakage.

df['moving_avg_4'] = df['Weekly_Sales'].transform(lambda x: x.shift(1).rolling(4).mean())
df['moving_avg_13'] = df['Weekly_Sales'].transform(lambda x: x.shift(1).rolling(13).mean())

df.head(15)
```

## Benchmark performance

We'll split the data and evaluate the benchmark models on the test set.

For these benchmark models, splitting the data into train and test is not so important. 

But we want to use the same procedure that we will use for machine learning models in the next notebook, so that we can compare the results.

```python
from sklearn.metrics import mean_squared_error
import numpy as np
```

```python

#Drop rows with NaNs (VERY IMPORTANT)

df_eval = df.dropna(subset=[
    'Weekly_Sales',
    'naive_pred',
    'lag2',
    'lag3',
    'lag4',
    'lag13',
    'moving_avg_4',
    'moving_avg_13',
    'seasonal_naive_pred'
])
```

```python
# Split into train and test

# We leave 2012 in the test set

# set date as index
df = df.set_index('Date')

split_date = '2012-01-01'

X_train = df[df.index < split_date]
X_test  = df[df.index >= split_date]
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
# The different basic forecasts

df.columns
```

```python
# Determine the error in the test set for each forecast
for column in df.columns[1:]:    
    target = X_test["Weekly_Sales"]
    preds = X_test[column]

    rmse = np.sqrt(mean_squared_error(target, preds))

    print(f"performance of {column} = {np.round(rmse, 0)}")
```

From these basic forecasts, the **naive forecast** is the best for this particular time series. 

We could of course try different seasonal patterns or windows for the moving average, but I'll leave that with you.

## Plot benchmark predictions

We will now overlay the predictions over the real values of the time series.

```python


train_sample = df[df.index < split_date]
test_sample  = df[df.index >= split_date]
fig, ax = plt.subplots(figsize=(12,5))

import matplotlib.ticker as mtick
plt.gca().yaxis.set_major_formatter(mtick.StrMethodFormatter('{x:,.0f}'))

train_sample['Weekly_Sales'].plot(ax=ax, label='train', linewidth=2, alpha=0.7)
test_sample['Weekly_Sales'].plot(ax=ax, label='test', linewidth=2, color='orange')
test_sample['naive_pred'].plot(ax=ax, label='prediction', linestyle='--', linewidth=2, color='green')

plt.xlabel('Date')
plt.ylabel('Weekly Sales')
ax.legend()
ax.grid(True)
plt.title(f'Naive Forecast – Global Univariate')
```

Fast but noisy

Too many points, we can't see anything. 

Let's limit the plot to 100 points:

```python

train_sample = df[df.index < split_date]
test_sample  = df[df.index >= split_date]
fig, ax = plt.subplots(figsize=(12,5))

train_sample.tail(50)['Weekly_Sales'].plot(ax=ax, label='train', linewidth=2)
test_sample.head(50)['Weekly_Sales'].plot(ax=ax, label='test', linewidth=2)
test_sample.head(50)['seasonal_naive_pred'].plot(ax=ax, label='prediction', linestyle='--', linewidth=2)

plt.xlabel('Date')
plt.ylabel('Weekly Sales')
ax.legend()
plt.title(f'Seasonal Naive Forecast – Global Univariate')
```

Captures yearly pattern

```python
train_sample = df[df.index < split_date]
test_sample  = df[df.index >= split_date]
fig, ax = plt.subplots(figsize=(12,5))

import matplotlib.ticker as mtick
plt.gca().yaxis.set_major_formatter(mtick.StrMethodFormatter('{x:,.0f}'))

train_sample.tail(50)['Weekly_Sales'].plot(ax=ax, label='train', linewidth=2, alpha=0.7)
test_sample.head(50)['Weekly_Sales'].plot(ax=ax, label='test', linewidth=2, color='orange')
test_sample.head(50)['moving_avg_4'].plot(ax=ax, label='prediction', linestyle='--', linewidth=2, color='green')

plt.xlabel('Date')
plt.ylabel('Weekly Sales')
ax.legend()
ax.grid(True)
plt.title(f'Moving Average Forecast – Global Univariate')
```

Smooth but lagging

```python
import matplotlib.pyplot as plt

sample = test_sample

plt.figure(figsize=(12,6))

import matplotlib.ticker as mtick
plt.gca().yaxis.set_major_formatter(mtick.StrMethodFormatter('{x:,.0f}'))

plt.plot(sample['Weekly_Sales'], label='Actual')
plt.plot(sample['naive_pred'], label='Naive')
plt.plot(sample['moving_avg_4'], label='Moving Avg')
plt.plot(sample['seasonal_naive_pred'], label='Seasonal Naive')
plt.xlabel('Date')
plt.ylabel('Weekly Sales')
plt.legend()
plt.title("Baseline Model Comparison")
plt.show()
```

Observation: Naive forecasting reacts quickly but produces volatile predictions.
Moving average smooths noise but fails to capture sharp demand spikes.
Seasonal naive captures recurring yearly patterns but struggles with irregular fluctuations.
This highlights the need for machine learning models that can capture both seasonality and dynamic changes.

Conclusion: Among baseline models, seasonal naive forecasting (52-week lag) provides the most meaningful benchmark, as it captures recurring yearly demand patterns. While naive and moving average models either overreact or oversmooth the data, seasonal naive aligns more closely with observed seasonal peaks, making it the strongest baseline for comparison.”
