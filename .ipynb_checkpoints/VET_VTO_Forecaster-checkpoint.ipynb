```python
# pip install flask pandas joblib

from flask import Flask, request, jsonify
import pandas as pd
import joblib
import numpy as np

app = Flask(__name__)
```

```python
# ---------------------------------------------------
# LOAD SAVED FORECASTER
# ---------------------------------------------------

bundle = joblib.load("warehouse_system.pkl")

forecaster = bundle["forecaster"]
vet_threshold = bundle["vet_threshold"]
vto_threshold = bundle["vto_threshold"]

```

```python
# ---------------------------------------------------
# SETTINGS
# ---------------------------------------------------

WORKERS_PER_UNIT = 25000        # demand per worker
LABOR_COST_PER_WORKER = 900     # optional cost estimate

# ---------------------------------------------------
# ROOT TEST
# ---------------------------------------------------

@app.route("/")
def home():
    return {"message": "Warehouse Forecast API Running"}
```

```python
# ---------------------------------------------------
# FORECAST ENDPOINT
# ---------------------------------------------------

@app.route("/forecast", methods=["POST"])
def forecast():

    data = request.json

    # -----------------------------------------
    # Input JSON Example:
    # {
    #   "weeks":12,
    #   "temperature":[42,45,40...],
    #   "fuel_price":[3.2,3.1...],
    #   "cpi":[220,221...],
    #   "unemployment":[7.1,7.0...],
    #   "isholiday":[0,0,0,1...]
    # }
    # -----------------------------------------

    weeks = data.get("weeks", 12)

    future_exog = pd.DataFrame({
        "IsHoliday": data["isholiday"][:weeks],
        "Temperature": data["temperature"][:weeks],
        "Fuel_Price": data["fuel_price"][:weeks],
        "CPI": data["cpi"][:weeks],
        "Unemployment": data["unemployment"][:weeks],
        "sales_velocity": 0,
        "backlog_proxy": 0,
        "extreme_temp": 0
    })

    future_exog["extreme_temp"] = (
        (future_exog["Temperature"] > 90) |
        (future_exog["Temperature"] < 32)
    ).astype(int)

    future_dates = pd.date_range(
        start="2012-01-06",
        periods=weeks,
        freq="W-FRI"
    )

    future_exog.index = future_dates

    # -----------------------------------------
    # FORECAST
    # -----------------------------------------

    pred = forecaster.predict(
        steps=weeks,
        exog=future_exog
    )

    output = []

    cumulative_cost = 0

    for i, value in enumerate(pred):

        # ---------------------------
        # Decision Logic
        # ---------------------------

        if value >= vet_threshold:
            decision = "VET"
            extra_workers = int((value - vet_threshold) / WORKERS_PER_UNIT)
            reduce_workers = 0
            est_cost = extra_workers * LABOR_COST_PER_WORKER

        elif value <= vto_threshold:
            decision = "VTO"
            extra_workers = 0
            reduce_workers = int((vto_threshold - value) / WORKERS_PER_UNIT)
            est_cost = -reduce_workers * LABOR_COST_PER_WORKER

        else:
            decision = "NORMAL"
            extra_workers = 0
            reduce_workers = 0
            est_cost = 0

        cumulative_cost += est_cost

        output.append({
            "week": i + 1,
            "predicted_demand": round(float(value), 2),
            "decision": decision,
            "extra_workers_needed": extra_workers,
            "workers_to_reduce": reduce_workers,
            "estimated_cost": est_cost,
            "cumulative_future_cost": cumulative_cost
        })

    return jsonify(output)

# ---------------------------------------------------
# RUN
# ---------------------------------------------------

if __name__ == "__main__":
    app.run(debug=True)
```
